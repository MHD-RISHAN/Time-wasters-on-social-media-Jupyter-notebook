# Time-Wasters on Social Media

 Social media has become an essential part of modern life, but it often leads to excessive time consumption and reduced productivity.This project explores how different platforms contribute to time-wasting habits among users of various age groups and professions.

By analyzing user behavior, engagement patterns, and addiction levels, we identify key factors driving unproductive screen time.
The goal is to uncover insights that can help users manage their usage more mindfully and reduce productivity loss.

# Plots

```python
df['Age'].value_counts().head(10).plot(kind='barh',color='k',edgecolor='red')
plt.title('Age wise classification',fontsize=18,fontname='Times New Roman',fontstyle='italic')
plt.show()
```

```python
plt.figure(figsize=(10,5))
df['Gender'].value_counts().plot(kind='pie',colors=['Grey','pink','skyblue'],explode=(0,0,0.1),shadow=True,autopct='%1.2f%%',startangle=60)
plt.title('Gender wise Analysis',fontsize=18,fontname='Times New Roman',fontstyle='italic')
plt.show()
```

```python
plt.figure(figsize=(8,5))
df['Location'].value_counts().plot(kind='bar',color='darkred',edgecolor='k')
plt.title('Analysis based on Locations',fontsize=18,fontname='Times New Roman',fontstyle='italic')
plt.show()
```

> in this plot shows India is first in Time-waste in social meadia and germany is Last .

```python
plt.figure(figsize=(8,5))
df.groupby('Profession')['Time Spent On Video'].count().sort_values(ascending=False).plot(kind='line',color='blue',marker='o',markerfacecolor='k',markeredgecolor='red')
plt.xticks(rotation=90)
plt.title('Analysis Based on Profession',fontsize=18,fontname='Times New Roman',fontstyle='italic')
plt.show()
```

> On the above plot defines Students and waiting staff are more time time sprnd in social media and teaching proffessional are less time of usage 

```python
plt.figure(figsize=(8,5))
sns.lineplot(x='Platform',y='Time Spent On Video',data=df,color='red',marker='o',markerfacecolor='k',markeredgecolor='yellow')
plt.title('Analysis Based on Platform',fontsize=18,fontname='Times New Roman',fontstyle='italic')
plt.xlabel('PLATFORMS',color='blue',fontsize=11,fontname='Times New Roman',fontstyle='italic')
plt.ylabel('TIMES SPEND ON VEDIO',color='blue',fontsize=11,fontname='Times New Roman',fontstyle='italic')
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.barplot(x='Demographics',y='Addiction Level',data=df,hue='Watch Reason',palette='Greens')
plt.title('Analysis Based on Addiction level & Reason',fontsize=18,fontname='Times New Roman',fontstyle='italic')
plt.xlabel('Demographics',color='red',fontsize=11,fontname='Times New Roman',fontstyle='italic')
plt.ylabel('Addiction Level',color='red',fontsize=11,fontname='Times New Roman',fontstyle='italic')
plt.show()
```

# Overall Findings

* Gender Usage Pattern:

    Males tend to spend more time on social media than females and show higher levels of addiction.

* Country-wise Comparison:

    India ranks highest in time wasted on social media, while Germany shows the lowest usage.

* Profession-Based Usage:

    Students and waiting staff are the heaviest users of social media, whereas teaching professionals have the lowest usage rates.

* Platform Preference:

    YouTube is the most widely used platform for time-wasting activities across users.

* Rural vs Urban Behavior:

    Users in rural areas mainly use social media due to procrastination, while in urban areas, it has become a habitual behavior.

* Content Importance:

    The majority of users give the highest importance to watching jokes and memes, indicating a strong preference for light and entertaining content.

* Age-Based Addiction:

    Young users are significantly more addicted to social media compared to middle-aged and senior individuals.
